#**1. Erosion**
#Properties:

Erosion removes pixels from the boundaries of objects, effectively shrinking them.
It is useful for removing small noise and separating objects that are close to each other.
#Code Implementation:

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image
image = cv2.imread('/content/bianry image with Noise.jpeg', cv2.IMREAD_GRAYSCALE)

# Define the structuring element
kernel = np.ones((5, 5), np.uint8)

# Apply erosion for 3 iterations
erosion_1 = cv2.erode(image, kernel, iterations=1)
erosion_2 = cv2.erode(image, kernel, iterations=2)
erosion_3 = cv2.erode(image, kernel, iterations=3)

# Display the results
plt.figure(figsize=(18, 6))
plt.subplot(1, 4, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(erosion_1, cmap='gray')
plt.title('Erosion - Iteration 1')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(erosion_2, cmap='gray')
plt.title('Erosion - Iteration 2')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(erosion_3, cmap='gray')
plt.title('Erosion - Iteration 3')
plt.axis('off')

plt.show()
```

**Analysis:**

As the number of iterations increases, the objects in the image shrink more, and small noise is removed.
Erosion is effective for separating objects that are close to each other.


#**2. Dilation**
# Properties:

Dilation adds pixels to the boundaries of objects, effectively expanding them.
It is useful for filling small holes and connecting nearby objects.

```python
# Apply dilation for 3 iterations
dilation_1 = cv2.dilate(image, kernel, iterations=1)
dilation_2 = cv2.dilate(image, kernel, iterations=2)
dilation_3 = cv2.dilate(image, kernel, iterations=3)

# Display the results
plt.figure(figsize=(18, 6))
plt.subplot(1, 4, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(dilation_1, cmap='gray')
plt.title('Dilation - Iteration 1')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(dilation_2, cmap='gray')
plt.title('Dilation - Iteration 2')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(dilation_3, cmap='gray')
plt.title('Dilation - Iteration 3')
plt.axis('off')

plt.show()
```

**Analysis**

As the number of iterations increases, the objects in the image expand more, and small holes are filled.
Dilation is effective for connecting nearby objects and filling small gaps.


# **3. Opening Properties:**

Opening is a combination of erosion followed by dilation.
It is useful for removing small noise and preserving the shape of larger objects.

```python
# Apply opening for 3 iterations
opening_1 = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel, iterations=1)
opening_2 = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel, iterations=2)
opening_3 = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel, iterations=3)

# Display the results
plt.figure(figsize=(18, 6))
plt.subplot(1, 4, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(opening_1, cmap='gray')
plt.title('Opening - Iteration 1')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(opening_2, cmap='gray')
plt.title('Opening - Iteration 2')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(opening_3, cmap='gray')
plt.title('Opening - Iteration 3')
plt.axis('off')

plt.show()
```

# **Analysis:**

Opening removes small noise and preserves the shape of larger objects.
As the number of iterations increases, more noise is removed, but the shape of larger objects remains intact.

# **4. Closing**
# **Properties:**

Closing is a combination of dilation followed by erosion.
It is useful for filling small holes and connecting nearby objects.

```python
# Apply closing for 3 iterations
closing_1 = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel, iterations=1)
closing_2 = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel, iterations=2)
closing_3 = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel, iterations=3)

# Display the results
plt.figure(figsize=(18, 6))
plt.subplot(1, 4, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(closing_1, cmap='gray')
plt.title('Closing - Iteration 1')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(closing_2, cmap='gray')
plt.title('Closing - Iteration 2')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(closing_3, cmap='gray')
plt.title('Closing - Iteration 3')
plt.axis('off')

plt.show()
```

#Analysis:

Closing fills small holes and connects nearby objects.
As the number of iterations increases, more holes are filled, and nearby objects are connected.

#**Comparison of Results**
#**Erosion vs. Dilation:**

Erosion shrinks objects and removes small noise, while dilation expands objects and fills small holes.
Erosion is useful for separating objects, while dilation is useful for connecting objects.


#**Opening vs. Closing:**

Opening removes small noise and preserves the shape of larger objects, while closing fills small holes and connects nearby objects.
Opening is useful for noise removal, while closing is useful for filling gaps.

#**Effect of Iterations:**

Increasing the number of iterations amplifies the effects of each operation.
For erosion and opening, more iterations result in more noise removal.
For dilation and closing, more iterations result in more filling of holes and connecting of objects.



# **Documentation**
# **Erosion:**

Effective for removing small noise and separating objects.
Challenges: May shrink objects too much if too many iterations are used.
# **Dilation:**

Effective for filling small holes and connecting objects.
Challenges: May expand objects too much if too many iterations are used.
# **Opening:**

Effective for noise removal while preserving the shape of larger objects.
Challenges: May remove too much noise if too many iterations are used.
# **Closing:**

Effective for filling small holes and connecting nearby objects.
Challenges: May fill too many holes if too many iterations are used

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the images
binary_image_with_noise = cv2.imread('/content/bianry image with Noise.jpeg', cv2.IMREAD_GRAYSCALE)
binary_image_with_gaps = cv2.imread('/content/binary image with gaps.png', cv2.IMREAD_GRAYSCALE)
grayscale_image = cv2.imread('/content/greyscale image of Mona Lisa.jpeg', cv2.IMREAD_GRAYSCALE)

# Print the image shapes to check if they loaded correctly
print(f"Shape of binary_image_with_noise: {binary_image_with_noise.shape if binary_image_with_noise is not None else None}")
print(f"Shape of binary_image_with_gaps: {binary_image_with_gaps.shape if binary_image_with_gaps is not None else None}")
print(f"Shape of grayscale_image: {grayscale_image.shape if grayscale_image is not None else None}")

# Define the structuring element
kernel = np.ones((5, 5), np.uint8)

# Apply erosion for 3 iterations on binary image with noise
erosion_1 = cv2.erode(binary_image_with_noise, kernel, iterations=1)
erosion_2 = cv2.erode(binary_image_with_noise, kernel, iterations=2)
erosion_3 = cv2.erode(binary_image_with_noise, kernel, iterations=3)

# Apply dilation for 3 iterations on binary image with gaps
dilation_1 = cv2.dilate(binary_image_with_gaps, kernel, iterations=1)
dilation_2 = cv2.dilate(binary_image_with_gaps, kernel, iterations=2)
dilation_3 = cv2.dilate(binary_image_with_gaps, kernel, iterations=3)

# Apply opening for 3 iterations on grayscale image
opening_1 = cv2.morphologyEx(grayscale_image, cv2.MORPH_OPEN, kernel, iterations=1)
opening_2 = cv2.morphologyEx(grayscale_image, cv2.MORPH_OPEN, kernel, iterations=2)
opening_3 = cv2.morphologyEx(grayscale_image, cv2.MORPH_OPEN, kernel, iterations=3)

# Apply closing for 3 iterations on grayscale image
closing_1 = cv2.morphologyEx(grayscale_image, cv2.MORPH_CLOSE, kernel, iterations=1)
closing_2 = cv2.morphologyEx(grayscale_image, cv2.MORPH_CLOSE, kernel, iterations=2)
closing_3 = cv2.morphologyEx(grayscale_image, cv2.MORPH_CLOSE, kernel, iterations=3)

# Display the results for erosion
plt.figure(figsize=(18, 6))
plt.subplot(1, 4, 1)
plt.imshow(binary_image_with_noise, cmap='gray')
plt.title('Original Image with Noise')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(erosion_1, cmap='gray')
plt.title('Erosion - Iteration 1')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(erosion_2, cmap='gray')
plt.title('Erosion - Iteration 2')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(erosion_3, cmap='gray')
plt.title('Erosion - Iteration 3')
plt.axis('off')

plt.show()

# Display the results for dilation
plt.figure(figsize=(18, 6))
plt.subplot(1, 4, 1)
plt.imshow(binary_image_with_gaps, cmap='gray')
plt.title('Original Image with Gaps')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(dilation_1, cmap='gray')
plt.title('Dilation - Iteration 1')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(dilation_2, cmap='gray')
plt.title('Dilation - Iteration 2')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(dilation_3, cmap='gray')
plt.title('Dilation - Iteration 3')
plt.axis('off')

plt.show()

# Display the results for opening
plt.figure(figsize=(18, 6))
plt.subplot(1, 4, 1)
plt.imshow(grayscale_image, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(opening_1, cmap='gray')
plt.title('Opening - Iteration 1')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(opening_2, cmap='gray')
plt.title('Opening - Iteration 2')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(opening_3, cmap='gray')
plt.title('Opening - Iteration 3')
plt.axis('off')

plt.show()

# Display the results for closing
plt.figure(figsize=(18, 6))
plt.subplot(1, 4, 1)
plt.imshow(grayscale_image, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(closing_1, cmap='gray')
plt.title('Closing - Iteration 1')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(closing_2, cmap='gray')
plt.title('Closing - Iteration 2')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(closing_3, cmap='gray')
plt.title('Closing - Iteration 3')
plt.axis('off')

plt.show()
```
